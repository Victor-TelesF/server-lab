```python
from domain.launch_config import LaunchConfig
from domain.trajectory_engine import TrajectoryEngine

import numpy as np
import matplotlib.pyplot as plt


config = LaunchConfig(v0=50, angle_deg=35, gravity=9.81)
engine = TrajectoryEngine(config)

times = np.linspace(0, engine.time_of_flight(), 100)

xs = [engine.horizontal_position(t) for t in times]
ys = [engine.vertical_position(t) for t in times]

plt.plot(xs, ys)
plt.xlabel("posição horizontal (m)")
plt.ylabel("posição vertical (m)")
plt.title("Trajetória do lançamento")
plt.grid(True)
plt.show()
```

```python
t_peak, height = engine.max_height()
print(f"Pico: t={t_peak:.2f}s, altura={height:.2f}m")

x_no_pico = engine.horizontal_position(t_peak)
plt.plot(xs, ys)
plt.scatter([x_no_pico], [height], color="red", zorder=5, label="altura máxima")
plt.legend()
plt.xlabel("posição horizontal (m)")
plt.ylabel("posição vertical (m)")
plt.title("Trajetória do lançamento")
plt.grid(True)
plt.show()
```
