```python
"""
Este arquivo tem valores ausentes de propósito.

Use df.isna().sum() para localizar os valores ausentes.
Preencha cada valor ausente com a média de sua respectiva coluna.
Converta as três colunas numéricas para uma matriz NumPy.
Calcule a média e o desvio-padrão de cada coluna usando axis=0.
Padronize os dados com (X - media) / desvio.
Confirme se as colunas padronizadas ficaram com média próxima de zero.
Desafio: encontre as linhas nas quais alguma medição padronizada tem valor absoluto maior que 1.5.
"""
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("exercicios_csv/leituras_sensores.csv")
print(df.shape)
print(df.dtypes)
print(df.head(10))

```

```python
colunas = ["temperatura_c", "umidade_pct", "pressao_hpa"]
df[colunas] = df[colunas].fillna(df[colunas].mean())



leituras = df[["temperatura_c", "umidade_pct", "pressao_hpa"]].to_numpy()
media = np.mean(leituras, axis=0)
desvio = np.std(leituras, axis=0)
dados_padronizados = (leituras - media) / desvio

print(dados_padronizados)
print(df.isna().sum())
print(dados_padronizados.mean(axis=0))
mascara = np.any(np.abs(dados_padronizados) > 1.5, axis=1)
print(df.loc[mascara])
```
