```python
import pandas as pd
import numpy as np

df = pd.read_csv("exercicios_csv/vendas_produtos.csv")
quantidades = df[["notebooks", "monitores", "teclados", "mouses"]].to_numpy()
precos = np.array([3500, 900, 180, 120])

print(quantidades.shape)
receita_total = quantidades @ precos
print(receita_total)
df["receita_total"] = receita_total
print(df.head(7))
indice_maior = df["receita_total"].idxmax()
maior_receita = df.loc[
    indice_maior,
    ["mes", "receita_total"]
    ]

print(maior_receita)
```

```python
precos_com_aumento = precos * 1.08
df["receita_total"] = quantidades @ precos_com_aumento
print(df.head())
```
