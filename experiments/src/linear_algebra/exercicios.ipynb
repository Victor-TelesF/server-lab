```python
import numpy as np

p0 = np.array([[2],
               [1]])

d1 = np.array([[3],
               [2]])

d2 = np.array([[-1],
               [ 4]])

d3 = np.array([[ 2],
               [-3]])

p_final = p0 + d1 + d2 + d3

deslocamento_total = d1 + d2 + d3

linha_reta = np.linalg.norm(p_final - p0)

distancia_total = (
    np.linalg.norm(d1)
    + np.linalg.norm(d2)
    + np.linalg.norm(d3)
)

print("Posição final:", p_final.flatten())
print("Deslocamento total:", deslocamento_total.flatten())
print("Comprimento da linha reta:", linha_reta)
print("Distância total percorrida:", distancia_total)
```
