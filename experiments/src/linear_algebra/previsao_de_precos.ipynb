```python
import pandas as pd
import numpy as np

df = pd.read_csv("exercicios_csv/casas.csv")
print(df.head(10))
print(df.shape)

x = df[["area_m2", "quartos", "idade_anos", "distancia_centro_km"]].to_numpy()
y = df["preco_real"].to_numpy()
pesos = np.array([2800, 18000, -900, -2500])
previsoes = x @ pesos
erros = y - previsoes
erro_absoluto_medio = np.mean(np.abs(erros))
print(previsoes)
print(erro_absoluto_medio)


```

```python
df["previsao"] = previsoes
df["erro"] = erros
df["erro_absoluto"] = np.abs(erros)

print(
    df[
        [
            "id",
            "preco_real",
            "previsao",
            "erro",
            "erro_absoluto",
        ]
    ]
)
```
