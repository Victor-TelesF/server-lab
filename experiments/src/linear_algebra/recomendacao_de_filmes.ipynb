```python
"""
Cada filme possui uma intensidade de 0 a 5 para cada gênero.

Crie uma matriz com acao, comedia, drama e ficcao.
Represente um usuário com preferencias = np.array([5, 1, 2, 4]).
Calcule a pontuação de todos os filmes usando matriz × vetor.
Adicione as pontuações ao DataFrame e ordene do maior para o menor.
Experimente outro vetor de preferências e observe como o ranking muda.
Desafio: divida a pontuação por np.linalg.norm(preferencias) e pesquise depois 
por que normalizações ajudam em sistemas de recomendação.
"""


```

```python


import pandas as pd
import numpy as np

df = pd.read_csv("exercicios_csv/filmes.csv")
print(df.head(8))
print(df.shape)

notas = df[["acao", "comedia", "drama", "ficcao"]].to_numpy()
usuario1 = np.array([5, 1, 2, 4])
usuario2 = np.array([2, 5, 5, 3])
preferencia = notas @ usuario1
pontuacao_usuario1 = pd.DataFrame({"filme": df["filme"], "pontuacao": preferencia})
pontuacao_usuario1 = pontuacao_usuario1.sort_values(by="pontuacao", ascending=False)
print(pontuacao_usuario1)

normalizacao1 = np.linalg.norm(usuario1)
preferencia_normalizada = usuario1 / normalizacao1
print(f"Preferencias normalizadas: {preferencia_normalizada}")


```

```python

pontuacao_usuario2 = pd.DataFrame({"filme": df["filme"], "pontuacao": notas @ usuario2})
pontuacao_usuario2 = pontuacao_usuario2.sort_values(by="pontuacao", ascending=False)
print(pontuacao_usuario2)
normalizacao2 = np.linalg.norm(usuario2)
preferencia_normalizada2 = usuario2 / normalizacao2
print(f"Preferencias normalizadas: {preferencia_normalizada2}")
```
