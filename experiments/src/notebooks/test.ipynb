```python
import numpy as np
import matplotlib.pyplot as plt

# 3 pontos de um triângulo (cada coluna é um ponto: x na linha 0, y na linha 1)
triangulo = np.array([
    [0, 2, 1],   # coordenadas x dos 3 pontos
    [0, 0, 2]    # coordenadas y dos 3 pontos
])

def plotar(pontos, cor, label):
    # fecha a forma repetindo o primeiro ponto no final
    x = list(pontos[0]) + [pontos[0][0]]
    y = list(pontos[1]) + [pontos[1][0]]
    plt.plot(x, y, marker='o', color=cor, label=label)

plt.figure(figsize=(6,6))
plotar(triangulo, 'blue', 'original')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.legend()
plt.axis('equal')
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.show()

# matriz de rotação de 90°
R = np.array([
    [0, -1],
    [1,  0]
])

triangulo_rotacionado = R @ triangulo

plt.figure(figsize=(6,6))
plotar(triangulo, 'blue', 'original')
plotar(triangulo_rotacionado, 'red', 'rotacionado 90°')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.legend()
plt.axis('equal')
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.show()

D = np.array([
    [2, 0],
    [0, 2]
])

triangulo_dilatado = D @ triangulo

plt.figure(figsize=(6,6))
plotar(triangulo, 'blue', 'original')
plotar(triangulo_dilatado, 'green', 'dilatado 2x')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.legend()
plt.axis('equal')
plt.xlim(-4, 6)
plt.ylim(-4, 6)
plt.show()

F = np.array([
    [-1, 0],
    [ 0, 1]
])

triangulo_refletido = F @ triangulo

plt.figure(figsize=(6,6))
plotar(triangulo, 'blue', 'original')
plotar(triangulo_refletido, 'purple', 'refletido (eixo y)')
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.legend()
plt.axis('equal')
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.show()
```
