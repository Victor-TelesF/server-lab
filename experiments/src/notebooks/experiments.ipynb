```python
l1 = [2,4,3]
l2 = [5,6,4]
def sum_list(l1: list, l2: list) -> list:
    l1.reverse()
    l2.reverse()
    s = str(int("".join(map(str, l1))) + int("".join(map(str, l2))))
    l = [int(s[x]) for x in range(len(s))]
    return l

print(sum_list(l1,l2))
```

```python
nums = [3, 2 , 4]
target = 6

def twosum(nums: list[int], target: int) -> list[int]:
    value = 0
    ans = []
    for i in range(len(nums)):
        for k in range(len(nums)):
            if i != k:
                value = nums[i] + nums[k]
                if value == target:
                    ans.extend([i, k])
                    return ans

print(twosum(nums, target))
            
```
